## Data Cleaning
For **PUFC17_NATEM_VS1** (Nature of Employment), **PUFC18_PNWHRS_VS1** (Normal Working Hours per Day), **PUFC23_PCLASS_VS1** (Class of Worker (Primary Occupation)) 

## Import
Import **numpy** and **pandas**.

```python
import numpy as np
import pandas as pd
```

```python
# Loading the dataset into a DataFrame
labor_force_df = pd.read_csv("LFSPUFApril2016.CSV", dtype=pd.StringDtype())
```

```python
labor_force_df.info()
```

```python
#labor_force_df['PUFC17_NATEM'].unique()
#sentinel value is 'no work'
labor_force_df['PUFC17_NATEM'] = labor_force_df['PUFC17_NATEM'].replace({' ' : 'no work',
                                                         '1' : 'long term work',
                                                         '2' : 'short term work',
                                                         '3' : 'non union work'})

#sentinel value is '-1' to indicate no work, therefore no hours spent working
labor_force_df['PUFC18_PNWHRS'] = labor_force_df['PUFC18_PNWHRS'].replace({'  ' : '-1'})
#labor_force_df['PUFC18_PNWHRS'].unique()

#sentinel value is 'no work' indicating unemployed
labor_force_df['PUFC23_PCLASS'] = labor_force_df['PUFC23_PCLASS'].replace({' ' : 'no work',
                                                                           '0' : 'private household',
                                                                           '1' : 'private establishment',
                                                                           '2' : 'government',
                                                                           '3' : 'self employed',
                                                                           '4' : 'employer of own business',
                                                                           '5' : 'worked on own business',
                                                                           '6' : 'no pay from own business'})
labor_force_df['PUFC23_PCLASS'].unique()
```
